```python
from sklearn.preprocessing import LabelEncoder
```

```python
import pandas as pd
df=pd.read_csv('/content/student_performance_dataset.csv')
df.head()
```

```python
print("\n Missing Value")
print(df.isnull().sum())
```

```python
df.loc[0:5, 'Study_Hours_per_week']=None
df.loc[0:5,'Attendence_Rate']=None
df.loc[0:5,'Past_Exam']=None
df.loc[0:5,'Final_Exam']=None
print(df.isnull().sum()>0)
```

```python
df['Attendance_Rate'] = df['Attendance_Rate'].fillna(df['Attendance_Rate'].mean())
```

```python
df['Attendance_Rate'] = df['Attendance_Rate'].fillna(df['Attendance_Rate'].median())
```

```python
df['Attendance_Rate'] = df['Attendance_Rate'].ffill()
```

```python
df['Final_Exam'] = df['Final_Exam_Score'].fillna(df['Final_Exam_Score'].mean())
print("\nMissing values after imputation:")
print(df.isnull().sum())
```

```python
df = pd.get_dummies(df,columns=['Parental_Education_Level'])
print("\nOne-Hot Encoded colunms:")
print(df.columns.tolist())
```

```python
df['Study_Hours_per_week']=df['Study_Hours_per_week'].fillna(df['Study_Hours_per_week'].mean())
```

```python
le = LabelEncoder()
df['Pass_Fail_Encoder'] = le.fit_transform(df['Pass_Fail'])
print("\nLabel Encoder column")
print(df[['Pass_Fail','Pass_Fail_Encoder']].head())
```

```python
df['Gender_Encoder'] = le.fit_transform(df['Gender'])
df['Internet_Access_Encoder'] = le.fit_transform(df['Internet_Access_at_Home'])
df['Extracurricular_Encoder'] = le.fit_transform(df['Extracurricular_Activities'])
```

```python
df.to_csv("student_Performance_cleaned.csv", index=False)
print("\nCleaned dataset saved as 'student_performance_cleaned.csv'")
```
